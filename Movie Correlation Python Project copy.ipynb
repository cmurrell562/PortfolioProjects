```python
# Import Libraries

import pandas as pd
import seaborn as sns
import numpy as np

import matplotlib
import matplotlib.pyplot as plt
plt.style.use('ggplot')
from matplotlib.pyplot import figure

# Plot embedded within cell instead of pop-up window
%matplotlib inline 
matplotlib.rcParams['figure.figsize'] = (12,8) # Adjusts the configuration of the plots we will create

# Read in the data

df = pd.read_csv(r'/Users/cmurrell/Documents/python/Movies Python Project/movies.csv')
```

```python
df.head()
```

```python
# Let's see if there's any missing data by looping through every column

for col in df.columns:
    pct_missing = np.mean(df[col].isnull())
    print('{} - {}%'.format(col, pct_missing))

# Drop missing data
df = df.dropna()
```

```python
# Data types for our columns

df.dtypes
```

```python
# Convert budget and gross columns to get rid of .0s in budget

df['budget'] = df['budget'].astype('int64')
df['gross'] = df['gross'].astype('int64')

```

```python
df
```

```python
# Create a column to reflect the year only
df['year_correct'] = df['released'].str.split(',', n=1).str[1].str.strip().str[:4]

df
```

```python
# inplace = False does not change the original df
df = df.sort_values(by=['gross'], inplace=False, ascending = False)
# print(df_sorted.shape)
```

```python
#Change setting to show all rows and allow us to scroll
pd.set_option('display.max_rows', None)
```

```python
# Drop any duplicates

df.duplicated().sum()

# df['company'].drop_duplicates().sort_values(ascending =False) #Show distinct values
```

```python
# Hypothesis: Budget & company will have high correlation to gross
```

```python
# Scatter plot with budget vs gross

plt.scatter(x=df['budget'],y=df['gross'])

plt.title('Budget vs Gross Earnings')
plt.xlabel('Budget For Film')
plt.ylabel('Gross Earnings')

plt.show()
```

```python
df.head()
```

```python
#Plot budget vs gross using seaborn
#Regplot fits a regression line through the data
#Seems like postive corrlation

sns.regplot(x='budget', y='gross',data=df, scatter_kws={"color": "red"},line_kws={"color":"blue"})
```

```python
# Let's start looking at the correlation
```

```python
#Confirm high Correlatin between budget and gross from scatter plot above
df.corr() #pearson correlation method is default
```

```python
#Visualize correlation with heatmap
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True)

plt.title('Correlation Matrix for Numeric Features')
plt.xlabel('Movie Features')
plt.ylabel('Movie Features')


plt.show()
```

```python
df_numerized = df.copy()


for col_name in df_numerized.columns:
    if(df_numerized[col_name].dtype == 'object'):
        df_numerized[col_name] = df_numerized[col_name].astype('category')
        df_numerized[col_name] = df_numerized[col_name].cat.codes
df_numerized
```

```python
df.head()
```

```python
#Visualize correlation of all columns with heatmap, now that all columns are numeric
correlation_matrix = df_numerized.corr()
sns.heatmap(correlation_matrix, annot=True)

plt.title('Correlation Matrix for Numeric Features')
plt.xlabel('Movie Features')
plt.ylabel('Movie Features')


plt.show()
```

```python
correlation_mat = df_numerized.corr()

#convert square df (see above) into 2d matrix
# List view is easier to skim for corrlation sorting
corr_pairs = correlation_mat.unstack() 
sorted_pairs = corr_pairs.sort_values()

sorted_pairs
```

```python
high_corr = sorted_pairs[(sorted_pairs) > 0.5]
high_corr
```
